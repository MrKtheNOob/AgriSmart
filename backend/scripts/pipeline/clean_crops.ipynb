```python
def clean_requirement(requirement: dict):
    return {
        k: v for k, v in requirement.items()
        if isinstance(v, dict) and v.get("minimum") and v.get("maximum")
    }

    
ph = [crop["requirements"]["soil"]["ph"] for crop in data]
first_three = dict(list(ph[0].items())[:3])
print("avant",first_three)
r=clean_requirement(first_three)
print("après",r)

climate = [crop["requirements"]["soil"]["climate"] for crop in data]
first_three = dict(list(ph[0].items())[:3])
print("avant",first_three)
r=clean_requirement(first_three)
print("après",r)
    
```
